```python
#21. Create a function called count_vowels(text).
#The function should receive a string and return the number of vowels (a, e, i, o, u) in it.
#The function should treat uppercase and lowercase letters as the same.

def count_vowels(text):
    vowels=("a","e","i","o","u")
    count=0
    text=text.lower()
    for i in text:
      if i in vowels:
        count+=1
    return count
txt=input("Enter a world:")
res=count_vowels(txt)
print(res)
```

```python
#22. Create a function called remove_duplicates(items).
#The function should receive a list and return a new collection containing each element only once.
#Example: remove_duplicates([1, 2, 2, 3, 3, 4]) should return [1, 2, 3, 4] or an equivalent collection.

def remove_duplicates(items):
  unique_items=set(items)
  return list(unique_items)
res=remove_duplicates([1, 2, 2, 3, 3, 4])
print(res)
```

```python
#23. Create a function called common_elements(list1, list2).
#The function should receive two lists and return the elements that exist in both lists.
#Example: common_elements([1, 2, 3], [2, 3, 4]) should return [2, 3].

def common_elements(list1, list2):
  common_list=[]
  for i in list1:
    if i in list2:
      common_list.append(i)
  return common_list
res=common_elements([1, 2, 3], [2, 3, 4])
print(res)

```

```python
#24. Create a function called get_divisors(number).
#The function should receive a positive integer and return all numbers that divide it without a remainder.
#Example: get_divisors(12) should return [1, 2, 3, 4, 6, 12].

def get_divisors(number):
  divisors=[]
  for i in range(1,number+1):
    if number%i==0:
      divisors.append(i)
  return divisors
res=get_divisors(12)
print(res)
```

```python
#25. Create a function called check_result(grades).
#The function should receive a list of grades, calculate the average grade,
#and return "Pass" if the average is 60 or higher. Otherwise, return "Fail".

def check_result(grades):
  sum=0
  for i in grades:
    sum+=i
  avg=sum/len(grades)
  if avg>=60:
    return "Pass"
  else:
    return "Fail"
grade=[50,60,70]
res=check_result(grade)
print(res)
```

```python
#34. Create a simple shopping cart program. Store available products and their prices in a dictionary.
#Display the available products to the user.
#Ask the user to enter product names and add the selected products to a cart until the user enters "done".
#At the end, print all selected products and the total price. If the user enters a product that does not exist, print an error message.


products = {"Laptop": 15000, "Phone": 10000, "Mouse": 500, "Keyboard": 1000}
print("Available products:")
for product,price in products.items():
  print(product,price)
cart=[]
total_price=0
while True:
  product=input("Enter a product:")
  if product=="done":
    break
  elif product in products:
    cart.append(product)
    total_price+=products[product]
  else:
    print("Invalid product")
print("Selected products:",cart)
print("Total price:",total_price)
```

```python
#38. Given transactions = [100, -50, 200, -30, -20, 500],
#treat positive numbers as deposits and negative numbers as withdrawals.
#Calculate and print the final balance, total deposited amount, and total withdrawn amount. For this example, the final balance should be 700.

transactions = [100, -50, 200, -30, -20, 500]
total_deposits=0
total_withrawals=0
for i in transactions:
  if i>0:
    total_deposits+=i
  else:
    total_withrawals+=i
final_balance=total_deposits+total_withrawals
print("total_deposits:",total_deposits)
print("total_withrawals:",total_withrawals)
print("final_balance:",final_balance)
```

```python
#39. Create a simple ATM program using a while loop. Start with a balance of 5000.
#Display a menu with: 1- Deposit, 2- Withdraw, 3- Check Balance, 4- Exit. Deposit should increase the balance,
#Withdraw should decrease it only if there is enough money, Check Balance should display the current balance, and Exit should stop the program.

simple_ATM=5000
while True:
  print("1- Deposit")
  print("2- Withdraw")
  print("3- Check Balance")
  print("4- Exit")
  choice=int(input("Enter a choice:"))
  if choice==1:
    deposit=int(input("Enter a deposit:"))
    simple_ATM+=deposit
  elif choice==2:
    withdraw=int(input("Enter a withdraw:"))
    if withdraw<=simple_ATM:
      simple_ATM-=withdraw
    else:
      print("insufficient balance")
  elif choice==3:
    print("current balance:",simple_ATM)
  elif choice==4:
    break
  else:
    print("Invalid choice")
```
